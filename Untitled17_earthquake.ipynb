```python
class LeakageFreePreprocessor:

    def __init__(
        self,
        correlation_threshold=0.90,
        pca_variance=0.95
    ):

        self.correlation_threshold = (
            correlation_threshold
        )

        self.pca_variance = pca_variance

        self.imputer = None
        self.scaler = None
        self.pca = None

        self.feature_names_in_ = None
        self.selected_features_ = None
        self.removed_features_ = None

        self.lower_bounds_ = {}
        self.upper_bounds_ = {}

    def fit(self, X):

        X = X.copy()

        self.feature_names_in_ = (
            X.columns.tolist()
        )

        # --------------------------------------------------
        # 1. Median imputation fitted ONLY on training data
        # --------------------------------------------------

        self.imputer = SimpleImputer(
            strategy="median"
        )

        X_imp = pd.DataFrame(
            self.imputer.fit_transform(X),
            columns=X.columns,
            index=X.index
        )

        # --------------------------------------------------
        # 2. IQR boundaries fitted ONLY on training data
        # --------------------------------------------------

        X_clip = X_imp.copy()

        for col in X_clip.columns:

            Q1 = X_imp[col].quantile(0.25)

            Q3 = X_imp[col].quantile(0.75)

            IQR = Q3 - Q1

            lower = Q1 - 1.5 * IQR

            upper = Q3 + 1.5 * IQR

            self.lower_bounds_[col] = lower

            self.upper_bounds_[col] = upper

            X_clip[col] = X_clip[col].clip(
                lower=lower,
                upper=upper
            )

        # --------------------------------------------------
        # 3. Correlation filtering fitted ONLY on training
        # --------------------------------------------------

        corr_abs = X_clip.corr().abs()

        upper_triangle = corr_abs.where(
            np.triu(
                np.ones(
                    corr_abs.shape
                ),
                k=1
            ).astype(bool)
        )

        self.removed_features_ = [
            column
            for column in upper_triangle.columns
            if any(
                upper_triangle[column]
                > self.correlation_threshold
            )
        ]

        self.selected_features_ = [
            column
            for column in X_clip.columns
            if column not in self.removed_features_
        ]

        X_selected = X_clip[
            self.selected_features_
        ]

        # --------------------------------------------------
        # 4. Scaling fitted ONLY on training data
        # --------------------------------------------------

        self.scaler = StandardScaler()

        X_scaled = self.scaler.fit_transform(
            X_selected
        )

        # --------------------------------------------------
        # 5. PCA fitted ONLY on training data
        # --------------------------------------------------

        self.pca = PCA(
            n_components=self.pca_variance,
            svd_solver="full",
            random_state=RANDOM_STATE
        )

        self.pca.fit(
            X_scaled
        )

        return self

    def transform(self, X):

        X = X.copy()

        # Imputation
        X_imp = pd.DataFrame(
            self.imputer.transform(X),
            columns=self.feature_names_in_,
            index=X.index
        )

        # Outlier clipping
        X_clip = X_imp.copy()

        for col in self.feature_names_in_:

            X_clip[col] = X_clip[col].clip(
                lower=self.lower_bounds_[col],
                upper=self.upper_bounds_[col]
            )

        # Correlation-selected features
        X_selected = X_clip[
            self.selected_features_
        ]

        # Scaling
        X_scaled = self.scaler.transform(
            X_selected
        )

        # PCA
        X_pca = self.pca.transform(
            X_scaled
        )

        pca_columns = [
            f"PC{i+1}"
            for i in range(
                X_pca.shape[1]
            )
        ]

        return pd.DataFrame(
            X_pca,
            columns=pca_columns,
            index=X.index
        )

    def fit_transform(self, X):

        self.fit(X)

        return self.transform(X)
```

```python
classes = np.sort(
    y.unique()
)

def multiclass_specificity(
    y_true,
    y_pred
):

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=classes
    )

    specificities = []

    for i in range(
        len(classes)
    ):

        TP = cm[i, i]

        FN = (
            cm[i, :].sum()
            - TP
        )

        FP = (
            cm[:, i].sum()
            - TP
        )

        TN = (
            cm.sum()
            - TP
            - FN
            - FP
        )

        specificity = (
            TN / (TN + FP)
            if (TN + FP) > 0
            else 0
        )

        specificities.append(
            specificity
        )

    return np.mean(
        specificities
    )


def calculate_metrics(
    y_true,
    y_pred,
    y_prob
):

    y_true_bin = label_binarize(
        y_true,
        classes=classes
    )

    return {

        "Accuracy":
            accuracy_score(
                y_true,
                y_pred
            ),

        "Precision":
            precision_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "Sensitivity":
            recall_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "Specificity":
            multiclass_specificity(
                y_true,
                y_pred
            ),

        "F1-score":
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "ROC-AUC":
            roc_auc_score(
                y_true_bin,
                y_prob,
                multi_class="ovr",
                average="macro"
            ),

        "Balanced Accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred
            ),

        "Cohen Kappa":
            cohen_kappa_score(
                y_true,
                y_pred
            ),

        "MCC":
            matthews_corrcoef(
                y_true,
                y_pred
            ),

        "Log Loss":
            log_loss(
                y_true,
                y_prob,
                labels=classes
            )
    }
```

```python
cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=RANDOM_STATE
)

cv_results = []

fold_information = []

total_folds = TOTAL_VALIDATION_RUNS

# Define models dictionary
models = {
    "Logistic Regression": LogisticRegression(random_state=RANDOM_STATE, solver='lbfgs'), # Changed solver to 'lbfgs'
    "RBF-SVM": SVC(random_state=RANDOM_STATE, probability=True),
    "Random Forest": RandomForestClassifier(random_state=RANDOM_STATE),
    "Extra Trees": ExtraTreesClassifier(random_state=RANDOM_STATE),
    "XGBoost": XGBClassifier(random_state=RANDOM_STATE, use_label_encoder=False, eval_metric='mlogloss'),
    "LightGBM": LGBMClassifier(random_state=RANDOM_STATE),
    "CatBoost": CatBoostClassifier(random_state=RANDOM_STATE, verbose=0)
}

for run_number, (
    train_idx,
    val_idx
) in enumerate(
    cv.split(
        X_train_raw,
        y_train
    ),
    start=1
):

    print(
        f"\n{'=' * 70}"
    )

    print(
        f"VALIDATION RUN {run_number}/{total_folds}"
    )

    print(
        f"{'=' * 70}"
    )

    X_fold_train = (
        X_train_raw.iloc[train_idx]
        .copy()
    )

    X_fold_val = (
        X_train_raw.iloc[val_idx]
        .copy()
    )

    y_fold_train = (
        y_train.iloc[train_idx]
        .copy()
    )

    y_fold_val = (
        y_train.iloc[val_idx]
        .copy()
    )

    # --------------------------------------------------
    # Leakage-free preprocessing
    # --------------------------------------------------

    preprocessor = LeakageFreePreprocessor(
        correlation_threshold=
            CORRELATION_THRESHOLD,
        pca_variance=
            PCA_VARIANCE
    )

    X_fold_train_processed = (
        preprocessor.fit_transform(
            X_fold_train
        )
    )

    X_fold_val_processed = (
        preprocessor.transform(
            X_fold_val
        )
    )

    # --------------------------------------------------
    # SMOTE ONLY ON FOLD TRAINING DATA
    # --------------------------------------------------

    smote = SMOTE(
        random_state=
            RANDOM_STATE + run_number
    )

    X_fold_balanced, y_fold_balanced = (
        smote.fit_resample(
            X_fold_train_processed,
            y_fold_train
        )
    )

    fold_information.append({

        "Run": run_number,

        "Original Features":
            X_fold_train.shape[1],

        "After Correlation Filtering":
            len(
                preprocessor.selected_features_
            ),

        "PCA Components":
            X_fold_train_processed.shape[1],

        "Explained Variance":
            preprocessor.pca
            .explained_variance_ratio_
            .sum(),

        "Training Samples Before SMOTE":
            len(y_fold_train),

        "Training Samples After SMOTE":
            len(y_fold_balanced)
    })

    for model_name, model in models.items():

        current_model = clone(
            model
        )

        current_model.fit(
            X_fold_balanced,
            y_fold_balanced
        )

        y_val_pred = (
            current_model.predict(
                X_fold_val_processed
            )
        )

        y_val_prob = (
            current_model.predict_proba(
                X_fold_val_processed
            )
        )

        metrics = calculate_metrics(
            y_fold_val,
            y_val_pred,
            y_val_prob
        )

        metrics["Run"] = run_number

        metrics["Model"] = model_name

        cv_results.append(
            metrics
        )

cv_results_df = pd.DataFrame(
    cv_results
)

print(
    "\n20-run validation completed."
)

print(
    "Total validation observations:",
    len(cv_results_df)
)
```

```python
f1_pivot = cv_results_df.pivot(
    index="Run",
    columns="Model",
    values="F1-score"
)

model_order = list(
    models.keys()
)

f1_pivot = f1_pivot[
    model_order
]

friedman_stat, friedman_p = (
    friedmanchisquare(
        *[
            f1_pivot[model]
            for model in model_order
        ]
    )
)

print("=" * 70)
print("FRIEDMAN TEST")
print("=" * 70)

print(
    "Statistic:",
    round(friedman_stat, 6)
)

print(
    "p-value:",
    round(friedman_p, 6)
)

if friedman_p < 0.05:

    print(
        "Result: Significant difference exists among the models (p < 0.05)."
    )

else:

    print(
        "Result: No statistically significant overall difference detected (p >= 0.05)."
    )
```

```python
class_report = classification_report(
    y_test,
    best_test_pred,
    labels=classes,
    output_dict=True,
    zero_division=0
)

class_metrics = []

for class_label in classes:

    TP = cm[
        class_label,
        class_label
    ]

    FN = (
        cm[class_label, :].sum()
        - TP
    )

    FP = (
        cm[:, class_label].sum()
        - TP
    )

    TN = (
        cm.sum()
        - TP
        - FN
        - FP
    )

    sensitivity = (
        TP / (TP + FN)
        if TP + FN > 0
        else 0
    )

    specificity = (
        TN / (TN + FP)
        if TN + FP > 0
        else 0
    )

    precision = (
        TP / (TP + FP)
        if TP + FP > 0
        else 0
    )

    npv = (
        TN / (TN + FN)
        if TN + FN > 0
        else 0
    )

    f1 = class_report[
        str(class_label)
    ]["f1-score"]

    class_metrics.append({

        "Class":
            class_label,

        "Precision":
            precision,

        "Sensitivity":
            sensitivity,

        "Specificity":
            specificity,

        "F1-score":
            f1,

        "NPV":
            npv
    })

class_metrics_df = pd.DataFrame(
    class_metrics
)

display(
    class_metrics_df.round(4)
)
```

```python
def calculate_original_feature_importance(
    model,
    preprocessor,
    X_test_processed,
    y_test
):

    n_components = (
        X_test_processed.shape[1]
    )

    pc_names = [
        f"PC{i+1}"
        for i in range(
            n_components
        )
    ]

    # ---------------------------------------------
    # Case 1: Tree-based models
    # ---------------------------------------------

    if hasattr(
        model,
        "feature_importances_"
    ):

        pc_importance = (
            model.feature_importances_
        )

    # ---------------------------------------------
    # Case 2: Linear models
    # ---------------------------------------------

    elif hasattr(
        model,
        "coef_"
    ):

        coefficients = np.asarray(
            model.coef_
        )

        if coefficients.ndim == 2:

            pc_importance = np.mean(
                np.abs(coefficients),
                axis=0
            )

        else:

            pc_importance = (
                np.abs(coefficients)
            )

    # ---------------------------------------------
    # Case 3: Nonlinear SVM
    # ---------------------------------------------

    else:

        perm = permutation_importance(
            model,
            X_test_processed,
            y_test,
            scoring="f1_macro",
            n_repeats=10,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )

        pc_importance = (
            np.maximum(
                perm.importances_mean,
                0
            )
        )

    pc_importance = np.asarray(
        pc_importance
    )

    if len(pc_importance) != n_components:

        pc_importance = np.resize(
            pc_importance,
            n_components
        )

    # ---------------------------------------------
    # Map PCA importance back to original features
    # ---------------------------------------------

    loading_matrix = np.abs(
        preprocessor.pca.components_.T
    )

    original_importance = (
        loading_matrix
        @ pc_importance
    )

    original_importance = pd.Series(
        original_importance,
        index=preprocessor.selected_features_
    )

    if original_importance.sum() > 0:

        original_importance = (
            original_importance
            / original_importance.sum()
        )

    return (
        original_importance
        .sort_values(
            ascending=False
        )
    )


original_importance = (
    calculate_original_feature_importance(
        BEST_MODEL,
        final_preprocessor,
        X_test_processed,
        y_test
    )
)

display(
    original_importance
    .to_frame(
        "Relative Importance"
    )
    .round(4)
)
```
